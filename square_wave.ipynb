```python
import matplotlib.pyplot as plt
import numpy as np

def plot_function(func, x_range):
  """
  Plots a given function within a specified x-range.

  Args:
    func: The function to plot (a Python function).
    x_range: A tuple or list representing the range of x-values (e.g., (-5, 5)).
  """
  x = np.linspace(x_range[0], x_range[1], 400) # Generate x values
  y = func(x) # Calculate corresponding y values

  plt.figure(figsize=(8, 6)) # Set figure size
  plt.plot(x, y) # Plot the function
  plt.xlabel("x") # Label x-axis
  plt.ylabel("y") # Label y-axis
  plt.title(f"Plot of y = {func.__name__}(x)") # Set title
  plt.grid(True) # Add grid lines
  plt.show()


# Example usage of a sine function
def sine_function(x):
  ret = 0
  for i in range(1,1000,2):
    ret += np.sin(i*x) / i
  return ret

plot_function(sine_function, (-10, 10))

```
